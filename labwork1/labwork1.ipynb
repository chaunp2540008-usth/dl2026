```python
# Gradient Descent

def grad(fod, x, lr):
    return x - lr * fod(x)

# MSE
def f(x):
    return x * x


def parabolic_fod(x):
    return 2 * x


lr = 0.1
epsilon = 10e-3
x = 10
gd = 10
time = 0

while f(x) > epsilon:
    x = grad(parabolic_fod, x, lr)
    time += 1
    print(time, x, f(x))
```
